# Quickstart

Load a released CryptoVision model, identify a fish image at family, genus and species level, and see which pixels drove the prediction.

```python
from cryptovision.predict import CryptoVisionAI
from cryptovision.weights import class_names, load_pretrained

ARCH = "gated"                  # std | concat | gated | att (113 species), web_only | mixed (106 species)
IMAGE = "../images/lab_Eviota_albolineata_sample.jpg"   # example photo in the repo; use your own path

names = class_names(model=ARCH)
model = load_pretrained(ARCH)      # downloads the weights on first use
ai = CryptoVisionAI(model, names["family"], names["genus"], names["species"])
```

## Predict

```python
family, genus, species = ai.predict(IMAGE)
print(family, "|", genus, "|", species)
print("confidence (family, genus, species):", [round(float(c), 3) for c in ai.confidence])
```

## Saliency map

Bright regions are the pixels that influenced the species prediction most.

```python
saliency = ai.generate_saliency_map("species")
ai.plot_saliency_overlay(saliency)
```

## A folder of images

```python
from pathlib import Path

for path in sorted(Path(IMAGE).parent.glob("*.jp*g")):
    print(path.name, ai.predict(str(path)))
```
